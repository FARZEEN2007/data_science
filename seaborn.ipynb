```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data={
    "product":["laptop","phone","tablet","laptop","phone","tablet"],
    "sales":[50,80,40,70,90,60],
    "price":[50000,30000,20000,55000,32000,22000],
    "month":["jan","jan","feb","feb","mar","mar"]
}
sns.barplot(x="product",y="sales",data=data)                # 1. barplot
plt.show()                                      # compare sales of different products
```

```python
sns.histplot(x="sales",data=data)
plt.show()
```

```python
import numpy as np
data = np.random.randn(1000)
plt.figure(figsize=(10, 6))
sns.histplot(data, bins=30, kde=True, color='skyblue', edgecolor='black')
plt.title("distribution of random data using seaborn")
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.grid(axis='y',alpha=0.75)
plt.show()
```

```python
product = ["laptop","phone","tablet"]
sales = [80,50,30]

plt.pie(sales,labels=product,autopct='%4.2f%%')
plt.title("sales distribution")
plt.legend()
plt.show()
```

```python
x=[1,2,3,4,5]
y=[10,5,8,4,2]
plt.plot(x,y)
plt.fill_between(x,y)
plt.show
```

```python
mark = [42,50,52,55,60,62,65,90]
plt.boxplot(mark)
plt.show()
```

```python
sales= [
    [10,20,30,40,50],
    [15,25,35,45,55],
    [20,30,40,50,60]
]
sns.heatmap(sales,annot=True,cmap='Blues')
plt.show()
```

```python
#1st
product = ["laptop","phone","tablet","watch"]
sales = [50,80,40,70]

plt.bar(product,sales)
plt.title("sales distribution")
plt.xlabel("product")
plt.ylabel("sales")
plt.show()
```

```python
#2
days = ["mon","tue","wed","thu","fri"]
visitors = [120,120,180,160,120]

plt.plot(days,visitors)
plt.title("visitors distribution")
plt.xlabel("days")
plt.ylabel("visitors")
plt.show()
```

```python
#3
categories = ["food","travel","rent","shopping"]
expenses = [5000,3000,8000,4000]

plt.pie(expenses,labels=categories,autopct='%4.2f%%')
plt.title("expenses distribution")
plt.legend()
plt.show()
```

```python
#4
advertising = [10,20,30,40,50]
sales = [25,35,45,55,70]

plt.scatter(advertising,sales)
plt.title("advertising vs sales")
plt.xlabel("advertising")
plt.ylabel("sales")
plt.show()
```

```python
marks = [45,50,52,60,62,65,67,70,72,75,78,80,82,85,88,90,95]
bins=[40,50,60,70,80,90,100]
plt.figure(figsize=(10,6))
plt.hist(marks,bins=bins,edgecolor='black')
plt.title("marks distribution")
plt.xlabel("marks")
plt.ylabel("frequency")
plt.show()
```
