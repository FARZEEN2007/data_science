```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[2,4,6,8,10]
plt.plot(x,y)
plt.show()
```

```python
months=["Nov","Dec","Jan","Feb","Mar"]
sales=[40,30,75,80,35]
plt.plot(months,sales)
plt.show()
```

```python
months=["Nov","Dec","Jan","Feb","Mar"]
sales=[40,30,75,80,35]
profit=[50,60,20,100,70]
plt.plot(months,sales,label="Sales")     #name the label
plt.plot(months,profit,label="Profit")
plt.legend()                             #display the names
plt.show()
```

```python
months=["Nov","Dec","Jan","Feb","Mar"]
sales=[40,30,75,80,35]
plt.plot(months,sales)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.grid("True")
plt.plot(
    months,
    sales,
    marker="H",
    color="darkgreen",
    linestyle="--"
)
plt.show()
```

```python
months=["Nov","Dec","Jan","Feb","Mar"]
sales=[40,30,75,80,35]
plt.bar(months,sales)
plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly Sales")
plt.grid("True")
plt.plot(
    months,
    sales,
    marker="H",
    color="darkgreen",
    linestyle="--"
)
plt.show()
```

```python
study_hours=[1,2,3,4,5,6]
marks=[44,56,75,80,30,60]                    #scatter
plt.scatter(study_hours,marks)
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
marks=[30,35,40,50,55,65,
       70,76,80,85,90,95]                 #Histogram
plt.hist(marks)
plt.title("Marks Distribution")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
categories=["Food","Travel","Shopping","Others"]
amount=[6000,7000,10000,3000]                            #pie chart
plt.pie(amount,labels=categories)
plt.title("Monthly Expenses")
plt.show()
```
