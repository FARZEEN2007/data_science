```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/titanic.csv")
df
df.head()                # for 1st 5 rows
print(df.shape)           # show no of rows and colm
```

```python
df.info()            #  show colm name, data type
```

```python
df.describe()        # statistical summary(min,max etc)
```

```python
df.isnull().sum()                # show null values
```

```python
df["Embarked"]=df["Embarked"].fillna(
    df["Embarked"].mode()[0])
df
```

```python
print(df.isnull().sum())
print("before: ",df.shape[0])
```

```python
df=df.drop_duplicates()
```

```python
print("after: ",df.shape[0])
df=df.rename(columns={
    "Passengerid":"PassengerId",
    "2urvived":"Survived",
})
```

```python
df["Sex"]=df["Sex"].map({
    0:"Male",
    1:"Female",
})
df
```

```python
sns.histplot(data=df,x="Age",bins=20)

plt.title("Age distribution of titanic passengers")
plt.xlabel("Age")
plt.ylabel("Number of passengers")
plt.show()
```

```python
sns.countplot(data=df,x="Sex")

plt.title("Sex distribution of titanic passengers")
plt.xlabel("Sex")
plt.ylabel("Number of passengers")
plt.show()
```

```python
sns.countplot(data=df,x="Pclass",hue="Survived")

plt.title("Pclass distribution of titanic passengers")
plt.xlabel("Pclass")
plt.ylabel("Number of passengers")
plt.show()
```
