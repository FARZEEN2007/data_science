```python
import pandas as pd
data=[20,40,60,80,100]
s=pd.Series(data,index=['A','B','C','D','E'])
print(s)
```

```python
data={"NAME":"farzeen","age":19}
s=pd.Series(data)
print(s)
```

```python
data1=[10,30,50,70]
s=pd.Series(data1)
print(s.loc[1])
```

```python
data=[10,30,50,70]
s=pd.Series(data,index=['A','B','C','D'])
print(s.iloc[3])
```

```python
data2=[10,20,30,40]
s=pd.Series(data2,index=['A','B','C','D'])
print(s.loc["A":"B"])
print(s.iloc[:2])
print(s.loc[s>10])
```

```python
data3=[5,10,15,20]
s=pd.Series(data3)
print(s)
```

```python
sales = pd.Series([900,2000,3000,4000],index=["Monday","Tuesday","Wednesday","Thursday"])
print(sales)
print(sales.max())
print(sales.sum())
print(sales[sales<1000])
```

```python
data = [20,20,30,50,30,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.unique())
```

```python
data = [20,20,30,50,30,60,60,80]
ar=pd.Series(data)
print(ar)
print(ar.unique())
print(ar.nunique())                 #to get the count
print(ar.value_counts())            #counts of repeated numbers
```

```python
data = [20,20,30,None,50,30,60,60,80,None]
ar=pd.Series(data)
print(ar)
print(ar.isna())
```

```python
data = [20,20,30,None,50,30,60,60,80,None]
ar=pd.Series(data)
print(ar)
print(ar.notna())
```

```python
data = [20,20,30,None,50,30,60,60,80,None]
ar=pd.Series(data)
print(ar)
print(ar.dropna())       #USE FOR REMOVING NULL
```
